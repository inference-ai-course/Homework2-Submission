## 2: Check Core Packages (Week 1 carry-over)

````python
import subprocess
import importlib
print("=" * 60)
print("📦 CORE PACKAGES CHECK")
print("=" * 60)

core_packages = {
    'jupyter':     'Jupyter Notebook',
    'anthropic':   'Anthropic API (Claude)',
    'requests':    'HTTP requests',
    'dotenv':      'Environment variables (python-dotenv)',
    'pydantic':    'Data validation',
}

missing = []
for pkg, desc in core_packages.items():
    try:
        if importlib.util.find_spec(pkg):
        # __import__(pkg if pkg != 'dotenv' else 'dotenv')
            print(f"✅ {desc}")
    except ImportError:
        subprocess.check_call([sys.executable, "-m", "pip", "install", pkg])
        print(f"❌ {desc}  ← MISSING")
        missing.append(pkg)

print()
if missing:
    print(f"⚠️  Missing: {missing}")
    print("   Run: pip install -r requirements.txt")
else:
    print("✅ All core packages installed!")
````

## 3: Check Week 2 — Transformer & Tokenization Packages

````python
print("=" * 60)
print("🤖 TRANSFORMER & TOKENIZATION PACKAGES")
print("=" * 60)

ml_packages = {
    'torch':          'PyTorch (attention demos)',
    'transformers':   'HuggingFace Transformers (BertViz, LoRA)',
    'numpy':          'NumPy (attention math)',
    'matplotlib':     'Matplotlib (heatmaps)',
    'bertviz':        'BertViz (attention visualization)',
    'tiktoken':       'tiktoken (OpenAI/GPT tokenizer)',
    'sentencepiece':  'SentencePiece (T5/LLaMA tokenizer)',
}

import importlib.util
import subprocess
import sys

missing_ml = []
for pkg, desc in ml_packages.items():
    # 1. Use find_spec to check if it exists
    if importlib.util.find_spec(pkg):
        print(f"✅ {desc}")
    else:
        # 2. If it's NOT found, we attempt to install
        print(f"🔄 {desc} is missing. Installing...")
        try:
            # FIX: sys.executable is a property (no parens) and -m needs a dash
            subprocess.check_call([sys.executable, "-m", "pip", "install", pkg])
            print(f"✅ {desc} installed successfully!")
        except Exception as e:
            print(f"❌ Failed to install {pkg}: {e}")
            missing_ml.append(pkg)

print()
if missing_ml:
    print(f"⚠️  Still Missing: {missing_ml}")
    print("   Run: pip install -r requirements.txt")
else:
    print("✅ All ML/tokenization packages installed!")
````

````python
import shutil
import subprocess
from dotenv import load_dotenv
import os

print("=" * 60)
print("🔧 SYSTEM DEPENDENCIES CHECK")
print("=" * 60)

env_path = Path.cwd().parent / ".env"
load_dotenv(dotenv_path=env_path)
SUDO_PASSWORD = os.getenv("MY_SUDO_PASS")

dependencies = {
    'pdftoppm': 'poppler-utils',
    'tesseract': 'tesseract-ocr',
    'ffmpeg': 'ffmpeg'
}

def install_system_dependencies():
    # 1. Validation Check
    if not SUDO_PASSWORD:
        print("❌ Error: MY_SUDO_PASS not found in .env file.")
        return

    # 2. Pre-update (Optional but recommended)
    print("Refreshing package lists...")
    update_cmd = f"echo {SUDO_PASSWORD} | sudo -S apt update -y"
    subprocess.run(update_cmd, shell=True, capture_output=True)
    
    for command, package in dependencies.items():
        if shutil.which(command) is None:
            print(f"🔄 {command} not found. Installing {package}...")
            
            # 3. Use -y for 'yes' and -S for 'stdin password'
            install_cmd = f"echo {SUDO_PASSWORD} | sudo apt install -y {package}"
            
            try:
                # capture_output=True prevents your password/logs from flooding the console
                subprocess.run(install_cmd, shell=True, check=True, capture_output=True)
                print(f"✅ {package} installed successfully!")
            except subprocess.CalledProcessError as e:
                print(f"❌ Failed to install {package}. Check if your password is correct.")
        else:
            if command == 'pdftoppm':
                result = subprocess.run([command, '-v'], capture_output=True, text=True)
            else:result = subprocess.run([command, '--version'], capture_output=True, text=True)
            version_line = result.stderr.split('\n')[0] if result.stderr else result.stdout.split('\n')[0]
            print(f"✅ {command} found: {version_line}")
            
            

install_system_dependencies()
````

## 7: Check Ollama (Path B/C)
