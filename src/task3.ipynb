```python
from utils.yt_downloader import download_youtube_audio_v2
import os

data_dir = "data"
audio_dir = f"{data_dir}/audio"
downloads_dir = f"{audio_dir}/downloads"
```

```python

os.makedirs(audio_dir, exist_ok=True)
os.makedirs(downloads_dir, exist_ok=True)

# A playlist of short NLP talks
playlist_url = "https://www.youtube.com/playlist?list=PLoROMvodv4rOhcuXMZkNm7j3fVwBBY42z"
download_youtube_audio_v2(playlist_url, downloads_dir, is_playlist=True)
```

```python
from utils.transcriber import transcribe_folder_threaded

# Transcribe all audio files in the downloads folder using threads
# This will:
# 1. Process multiple files in parallel (default: 3 threads)
# 2. Save each transcription as a .txt file next to the audio file
# 3. Create a transcription_summary.jsonl file with all results

results = transcribe_folder_threaded(
    folder_path=downloads_dir,
    model_name="tiny",  # Use "base", "small", "medium", or "large" for better accuracy
    max_workers=3  # Adjust based on your CPU cores
)

# Print summary
print(f"\nTranscription Summary:")
print(f"Total files: {len(results)}")
print(f"Successful: {sum(1 for r in results if r['status'] == 'success')}")
print(f"Failed: {sum(1 for r in results if r['status'] != 'success')}")

# Show first few results
print("\nFirst 3 results:")
for result in results[:3]:
    print(f"\nFile: {result['file']}")
    print(f"Status: {result['status']}")
    print(f"Transcript saved to: {result.get('transcript_file', 'N/A')}")
    print(f"Preview: {result['transcription'][:100]}..." if result['transcription'] else "No transcription")
```
