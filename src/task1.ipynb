```python
import logging
from utils.arxiv_search import build_arxiv_search_url, run_scraper_in_background
from utils.arxiv_search import save_arxiv_scraped_details, scrape_arxiv_details_from_json_threaded

# get the arxiv url from a custom library
arxiv_url = build_arxiv_search_url(size=20, query="Artificial Intelligence")
ml_url = build_arxiv_search_url(size=20, query="Machine Learning")
neural_net_url = build_arxiv_search_url(size=20, query="Neural networks")
logging.debug(f"arxiv url: {arxiv_url} -> {neural_net_url} -> {ml_url}")
```

```python


run_scraper_in_background(url=arxiv_url, output_file="data/ai_arxiv_search.json")
run_scraper_in_background(url=ml_url, output_file="data/ml_arxiv_search.json")
run_scraper_in_background(url=neural_net_url, output_file="data/nn_arxiv_search.json")
```

```python
from utils.arxiv_search import scrape_arxiv_details_from_json_threaded, save_arxiv_scraped_details
arxiv_config = {
    "topics": {
        "AI": "Artificial Intelligence"
    },
    "page_sizes": [50],
    "file_locations": {
        "raw": "raw/",
        "clean": "clean/"
    }
}


from utils.arxiv_search import build_arxiv_search_url, run_scraper_in_background

for key, query in arxiv_config["topics"].items():
    for size in arxiv_config["page_sizes"]:
        url = build_arxiv_search_url(query=query, size=size)
        raw_path = f"{arxiv_config['file_locations']['raw']}arxiv_{key.lower()}_{size}.json"
        logging.debug(f"Running in background...")
        run_scraper_in_background(url, output_file=raw_path)
        logging.debug(f"Complete: {size}")
```

```python

for key in arxiv_config["topics"]:
    for size in arxiv_config["page_sizes"]:
        raw_path = f"{arxiv_config['file_locations']['raw']}arxiv_{key.lower()}_{size}.json"
        clean_path = f"{arxiv_config['file_locations']['clean']}arxiv_{key.lower()}_{size}_clean.json"
        details = scrape_arxiv_details_from_json_threaded(raw_path)
        save_arxiv_scraped_details(details, output_file=clean_path)

```
