```python
import logging
import os
from utils.arxiv_search import build_arxiv_search_url, run_scraper_in_background
from utils.arxiv_search import save_arxiv_scraped_details, scrape_arxiv_details_from_json_threaded

# get the arxiv url from a custom library
arxiv_url = build_arxiv_search_url(size=20, query="Artificial Intelligence")

# log it
logging.debug(f"arxiv url: {arxiv_url}")
```

```python
# create a folder
data_dir = "data"
scraped_data_dir = f"{data_dir}/scraped"
scraped_data_file = f"{scraped_data_dir}/scraped_data.json"

os.makedirs(data_dir, exist_ok=True)
os.makedirs(scraped_data_dir, exist_ok=True)

logging.debug(f"`{data_dir}` and `{scraped_data_dir}` Created!")

# run scrapper in background
logging.debug(f"Defined data file: {scraped_data_file }")
run_scraper_in_background(url=arxiv_url, output_file=scraped_data_file)
```

```python
from utils.arxiv_search import scrape_arxiv_details_from_json_threaded, save_arxiv_scraped_details

arxiv_details = scrape_arxiv_details_from_json_threaded(json_file=scraped_data_file, max_workers=4)

print(arxiv_details)
```

```python
# save scrapped details to output file
# scrapped_data_dir - this is the scrapped data dir
cleaned_data_dir = f"{data_dir}/cleaned"
# scrapped_details file

os.makedirs(cleaned_data_dir, exist_ok=True)

# cleaned file
cleaned_data_file = f"{cleaned_data_dir}/arxiv_clean.json"
# save the cleaned details
save_arxiv_scraped_details(results=arxiv_details, output_file=cleaned_data_file)
```

```python
# download pdf files
# pdf_dir
from utils.arxiv_search import get_pdf_arxiv

# define the pdf
pdf_dir = f"{data_dir}/pdf_ocr"

# create the dir
os.makedirs(pdf_dir, exist_ok=True)

# download pdf files
get_pdf_arxiv(cleaned_json=cleaned_data_file, save_dir=pdf_dir)
```
